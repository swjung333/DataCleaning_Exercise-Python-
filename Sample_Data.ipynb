### Cleaning Data Exercise: When we have missing values, outliers, inconsistent formatting, duplicates, invalid entries and etc.

```python
import pandas as pd
import numpy as np

# Generating artificial sample data

np.random.seed(42) 

n = 10000  

df = pd.DataFrame({
    "OrderID": np.random.randint(1000, 2000, n), 
    "CustomerID": np.random.randint(1, 500, n),
    "Age": np.random.choice([20, 25, 30, 35, 40, None, -10], n),
    "Price": np.random.choice([10, 20, 50, 100, None, 9999], n),  
    "City": np.random.choice(["Berlin", "berlin", "BERLIN", "Munich", "munich", None, "Munchen"], n),
    "OrderDate": np.random.choice([
        "2023-01-10", "10/02/2023", "March 5 2023", None
    ], n),
    "Product": np.random.choice(["Laptop", "Phone", "Tablet", "phone ", " LAPTOP"], n),
    "Quantity": np.random.choice([1, 2, 3, 5, None, -1], n),
    "Email": np.random.choice([
        "test@gmail.com", "user@yahoo.com", "invalid_email", None
    ], n)
})

df.to_csv("real_world_dirty_data.csv", index=False)
```

```python
print(df["Age"].unique())
```

```python
#clean "Age" column for non-numeric values
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Age"] = df["Age"].fillna(df["Age"].median()) #fillna is just to fill the missing values with the median of the column
```

```python
print(df["Age"].unique())
```

```python
#clean "Price"" column for non-numeric values
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
df["Price"] = df["Price"].fillna(df["Price"].median()) #fillna is just to fill the missing values with the median of the column
```

```python
print(df["Price"].unique())
```

```python
#check the initial unique values in the "Quantity" column
print(df["Quantity"].unique())
```

```python
#clean "Quantity" column for non-numeric values
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Quantity"] = df["Quantity"].fillna(0).astype(int) #fillna is just to fill the missing values with 0 and then convert the column to integer type
```

```python
print(df["Quantity"].unique())
```

```python
#check the initial unique values in the "City" column
print(df["City"].unique())
```

```python
df["City"] = df["City"].fillna("unknown") #in this time, fillna is just to fill the missing values with "unknown"
```

```python
#check the unique value in the "City" column after filling the missing values
print(df["City"].unique())
```

```python
#check the initial unique values in the "OrderID" column
print(df["OrderID"].unique())
```

```python
#removing duplicates
df = df.drop_duplicates(subset = "OrderID")
```

```python
print(df["Age"].unique())
```

```python
#abs function is used to get the absolute value of a number.
#abs(5) ===> 5
#abs(-5) ===> 5
#abs(0) ===> 0

df["Age"] = df["Age"].abs() #abs() function is used to convert negative values to positive values, and it leaves positive values and zero unchanged. In this case, it will convert any negative ages to their absolute values, ensuring that all ages are non-negative.
df["Age"] = df["Age"].apply(lambda x: abs(x)) # this is just another way to do the same thing as the previous above line.
```

```python
print(df["Age"].unique())
```

```python
#remove negative values from "Quantity" column.
df["Quantity"] = df["Quantity"].abs()
```

```python
print(df["Quantity"].unique())
```

```python
#check the original unique values in City column first
print(df["City"].unique())
```

```python
df["City"] = df["City"].str.lower().str.strip() #this removes leading and trailing spaces and converts all city names to lowercase.
df["City"] = df["City"].replace({"Munchen" : "munich"}) #using dictionary ({}) to replace "Munchen" with "munich"

df.loc[: , "City"] = df["City"] = df["City"].str.lower().str.strip() #This is just another way to do as the 1st line
df.loc[: , "City"] = df["City"] = df["City"].replace({"Munchen" : "munich"}) #This is just another way to do as the 2nd line
```

```python
print(df["Product"].unique())
```

```python
df["Product"] = df["Product"].str.lower().str.strip() #this removes leading and trailing spaces and converts all product names to lowercase.
```

```python
print(df["Product"].unique())
```

```python
#Fix Date (Standardize everything under "OrderDate" into one format)

df["OrderDate"] = pd.to_datetime(df["OrderDate"], errors = "coerce")

```

```python
print(df["OrderDate"].unique())
```

```python
#How to handle outliers: "Price" column has some outliers (e.g., 9999), we can remove it as it is not a realistic price for the products in our dataset.

df = df[df["Price"] <  1000] # remove rows where "Price" is greater than or equal to 1000
```

```python
print(df["Price"].unique())
```

```python
#How to validate email addresses: we can use a simple regular expression pattern to check if the email addresses are in a valid format. In other words, we can check if the email contains "@" and "." characters.

df = df[df["Email"].str.contains("@", na = False)]
```

```python
print(df["Email"].unique())
```

```python
# How to create "Revenue" column. (Revenue = Price * Quantity) why do we need revenue?
# Creating Revenue is for us to analyze the total revenue generated from all orders in our dataset. By creating a "Revenue" column, we can easily calculate the total revenue for each order and then aggregate it to get insights into the overall performance of our sales.
 
df["Revenue"] = df["Price"] * df["Quantity"] #create a new column "Revenue".
```
